### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
#Remove records with missing aircraft manufactured names
df = df.dropna(subset=["Make"]).copy()

#Standardize manufacturer names
df["Make"] = df["Make"].str.strip().str.upper()

#Count acciedent records for each manufacturer
make_counts = df["Make"].value_counts()

#Keep manufacturers with at least 50 accident records
valid_makes = make_counts[make_counts >= 50].index
df = df[df["Make"].isin(valid_makes)].copy()

#Verify the cleaned dataset
print("Dataset shape:", df.shape)
print("Number of manufacturers:", df["Make"].nunique())
print("\nManufacturer counts:")
print(df["Make"].value_counts().head(20))
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python
# Remove records with missing model names
df = df.dropna(subset=["Model"]).copy()

#Clean and standardize model names
df["Model"] = df["Model"].str.strip().str.upper()

#Check missing values values and dataset size
print("Missing Model values:", df["Model"].isnull().sum())
print("Dataset shape:", df.shape)

#Inspect the most common models
print("\nTop 20 aircraft models:")
print(df["Model"].value_counts().head(20))

#Check whether model names are shared across manufacturers
model_make_counts = df.groupby("Model")["Make"].nunique()

shared_models = model_make_counts[model_make_counts > 1]

print("\nNumber of model names shared by multiple manufacturers:")
print(len(shared_models))

print("\nExamples of shared moedl names:")
print(shared_models.head(20))

# Create a unique identifier for each aircraft type
df["Plane.Type"] = (
    df["Make"].str.strip().str.upper()
    + "_"
    +df["Model"]
)

#Display and verify
print("\nAircraft type examples:")
print(df[["Make", "Model", "Plane.Type"]].head(10))

print("\nNumber of unique aircraft types:", df["Plane.Type"].nunique())
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
# Columns to inspect an clean
columns_to_clean = [
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

# Inspect the original values and missing values
for column in columns_to_clean:
    print(f"\n--- {column} ---")
    print("Missing values:", df[column].isnull().sum())
    print(df[column].value_counts(dropna=False).head(20))

#Clean text columns by removing extra spaces and standadizing capitalisation
text_columns = [
    "Engine.Type",
    "Weather.Condition",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]    


for column in text_columns:
    df[column] = df[column].apply(
        lambda value: value.strip().upper()
        if isinstance(value, str) else value
    )

#Clean Number.of.Engines by converting it to Numeric
df["Number.of.Engines"] = pd.to_numeric(
    df["Number.of.Engines"], errors="coerce"
)    

#Verify the cleaned values
print("\n---Verfification after cleaning---")

for column in columns_to_clean:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False).head(15))

print("\nDataset shape:", df.shape)    
```

### Column Removal
